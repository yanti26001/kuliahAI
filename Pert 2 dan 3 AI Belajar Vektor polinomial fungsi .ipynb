```python
import numpy as np

#1. Eksperimen membuat vektor
#Vektor baris adalah aray 1 D
vector_a = np.array([1,2,3])
vector_b = np.array([4,5,6])

# 2. Aritmetika Dasar (Element-wise)
print(vector_a + vector_b) #Addition: [5,7,9]
print(vector_b - vector_a) #Subtraction: [3,3,3]
print(vector_a*2) #Scalar Multi: [2,4,6]
print(vector_a*vector_b) #Element-wise: [4,10,18]

# 3. Operasi matematis
dot_product = np.dot(vector_a,vector_b) #Dot Product: (1*4+2*5+3*6)=32
magnitude = np.linalg.norm(vector_a) #Magnitude/L2 Norm: sqrt(1^2 + 2^2 +3^2)
print(dot_product)
print(magnitude)
```

```python
import numpy as np

#1. Definisikan dua vektor 3D
#Vektor baris adalah aray 1 D
vector_a = np.array([1,2,3])
vector_b = np.array([4,5,6])

# 2. Hitung Cross Product menggunakan nm.cross()
cross_product=np.cross(vector_a,vector_b)

# 3. Tampilkan hasil
print("Vektor a :",vector_a)
print("Vektor b :",vector_b)
print("\nHasil Cross Product (axb):")  #\n artinya print di baris yang baru (enter)
print(cross_product)
```

```python
#Vector Outer Product
import numpy as np
# 1. Definisikan dua vektor
vector_a = np.array([1,2,3])
vector_b = np.array([4,5,6])

# 2.Hitung Outer Product menggunakan np.outer()
outer_product = np.outer(vector_a,vector_b)

# 3.Tampilkan hasil
print("Vektor a :",vector_a)
print("Vektor b :",vector_b)
print("\nHasil Outer Product (a X b) berukuran 3x3:")
print(outer_product)
```

```python
import numpy as np
#Operasi matrik-vektor
# 1.Tentukan sudut rotasi (misal 90 derajat atau pi/2 radian)
theta = np.pi/2
# 2.Buat Matriks Rotasi 2D (2 x 2) round itu membulatkan misal kecil sekali jadi 0
# Kalo mau ubah sudut yang lain ubah pi nya lalu round nya dihapus supaya hasilnya ngga pembulatan

R = np.round(np.array((
    [np.cos(theta),-np.sin(theta)],
    [np.sin(theta),np.cos(theta)]
)))
# 3. Buat Vektor Kolom v (2x1) menggunakan struktur array 2D, matriks berukuran 2 x 1, yaitu vektor kolom jadi ada dua kurung siku
v = np.array([[1], [0]])  #x nya 1 y nya 0

# 4. Perkalian Matriks dengan Vektor Kolom (R @ v)
v_rotated = np.round(R @ v)

# Tampilkan semua hasil ke terminal
print("Matriks Rotasi R (90 derajat):")
print(R)

print("\nVektor Kolom v awal:")
print(v)

print("\nHasil Rotasi (R @ v):")
print(v_rotated)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from numpy.polynomial import Polynomial as Poly

# 1. Definisikan polinomial P(x) = 1 + 2x + 3x^2
P = Poly([1, 2, 3])
# 2. Buat 100 titik x dari rentang 0 sampai 2 agar kurva mulus
x_plot = np.linspace(0, 2, 100)
# 3. Evaluasi nilai P(x) untuk ke-100 titik tersebut
y_plot = P(x_plot)
# 4. Proses pembuatan grafik menggunakan matplotlib
plt.figure(figsize=(8, 5))
plt.plot(x_plot, y_plot, label='$P(x) = 1 + 2x + 3x^2$', color='blue',
linewidth=2)
# Menambahkan titik-titik sampel (opsional, dari contoh 5 titik sebelumnya)
x_sampel = np.linspace(0, 2, 5)
y_sampel = P(x_sampel)
plt.scatter(x_sampel, y_sampel, color='red', zorder=5, label='Titik Sampel')
# Pengaturan dekorasi grafik
plt.title('Grafik Fungsi Polinomial P(x)', fontsize=14)
plt.xlabel('Nilai x', fontsize=12)
plt.ylabel('Nilai P(x)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=11)
# Menampilkan grafik ke layar
plt.show()
```

```python
import numpy as np

# Membuat rentang nilai x dari 0 sampai 2
x_piece = np.linspace(0, 2, 5) # [0. , 0.5, 1. , 1.5, 2. ]

# Tuliskan daftar kondisi (kondisi 1, kondisi 2)
kondisi = [x_piece < 1, x_piece >= 1]

# Tuliskan daftar fungsi yang bersesuaian berupa fungsi lambda
fungsi = [lambda x: x**2, lambda x: 2*x + 1]

# Evaluasi menggunakan np.piecewise
y_piece = np.piecewise(x_piece, kondisi, fungsi)

print("x   :", x_piece)
print("h(x) :", y_piece)
```

```python
import numpy as np
import matplotlib.pyplot as plt
# 1. Menyiapkan data x menggunakan np.linspace
# Mengambil rentang 0 sampai pi (sekitar 3.14) sebanyak 100 titik
x = np.linspace(0, np.pi, 100)
# 2. Mengevaluasi Fungsi Eksponensial: f(x) = e^(x)
# Sengaja menggunakan e^x (bukan e^2x) agar skala grafik tidak terlalu jomplang dengan sinus
y_exp = np.exp(x)
# 3. Mengevaluasi Fungsi Trigonometri: g(x) = 2 sin(x)
y_trig = 2 * np.sin(x)
# --- PROSES PLOTTING GRAFIK ---
plt.figure(figsize=(9, 5))
# Memplot kedua kurva ke dalam satu sumbu grafik yang sama
plt.plot(x, y_exp, label=r'$f(x) = e^x$', color='darkorange', linewidth=2.5)
plt.plot(x, y_trig, label=r'$g(x) = 2 \sin(x)$', color='teal', linewidth=2.5)
# Menambahkan dekorasi teks dan label grafik
plt.title('Gabungan Grafik Fungsi Eksponensial & Trigonometri', fontsize=14,
pad=15)
plt.xlabel('Nilai x (Radian)', fontsize=12)
plt.ylabel('Nilai Fungsi', fontsize=12)
# Mengatur batas tampilan sumbu X dan Y agar proporsional
plt.xlim(0, np.pi)
plt.ylim(0, np.max(y_exp) + 1)
# Mengaktifkan grid latar belakang dan legenda fungsi
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(fontsize=12, loc='upper left')
# Menampilkan grafik ke layar
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
# 1. Menyiapkan data x (0 sampai pi)
x = np.linspace(0, np.pi, 100)
# 2. Hitung nilai masing-masing fungsi
y_exp = np.exp(x)
y_trig = 2 * np.sin(x)
# 3. Membuat subplots: 1 baris, 2 kolom
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
# --- GRAFIK KIRI: Eksponensial ---
ax1.plot(x, y_exp, color='darkorange', linewidth=2.5, label=r'$f(x) = e^x$')
ax1.set_title('Fungsi Eksponensial', fontsize=12)
ax1.set_xlabel('Nilai x', fontsize=10)
ax1.set_ylabel('Nilai f(x)', fontsize=10)
ax1.grid(True, linestyle='--', alpha=0.5)
ax1.legend(loc='upper left')
# --- GRAFIK KANAN: Trigonometri ---
ax2.plot(x, y_trig, color='teal', linewidth=2.5, label=r'$g(x) = 2 \sin(x)$')
ax2.set_title('Fungsi Trigonometri', fontsize=12)
ax2.set_xlabel('Nilai x (Radian)', fontsize=10)
ax2.set_ylabel('Nilai g(x)', fontsize=10)
ax2.grid(True, linestyle='--', alpha=0.5)
ax2.legend(loc='upper right')
# Mengatur tata letakmlabel
plt.tight_layout()
# Menampilkan grafik
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Menyiapkan data waktu (t) dari 0 sampai 9.5 detik
t = np.linspace(0, 9.5, 500)

# 2. Menghitung nilai fungsi f(x) = e^(-0.5t) * sin(2t)
f_x = np.exp(-0.5 * t) * np.sin(2 * t)

# 3. Membuat figure dan plot
plt.figure(figsize=(7, 4.5))
plt.plot(t, f_x, label=r'$f(x) = e^{-0.5x}\sin(2x)$', color='darkorange', linewidth=2)

# 4. Pengaturan judul dan label sumbu
plt.title('Osilasi teredam', fontsize=11, pad=10)
plt.xlabel('t (Detik)', fontsize=10)
plt.ylabel('Nilai Fungsi', fontsize=10)

# 5. Mengatur batas sumbu X dan Y agar sesuai gambar
plt.xlim(0, 9.5)
plt.ylim(-1.9, 1.9)

# 6. Menambahkan grid dan legenda
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(loc='upper left', fontsize=9)

# 7. Menampilkan grafik
plt.tight_layout()
plt.show()
```
